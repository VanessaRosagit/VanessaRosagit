```python
# Criando as tuplas
mamiferos = ("cachorro", "gato", "elefante", "leão")
insetos = ("formiga", "abelha", "besouro")

# Criando uma terceira tupla a partir delas
animais = mamiferos + insetos
print("Tupla de animais:", animais)
# Tentando usar append -> vai dar erro
try:
    animais.append("tigre")
except AttributeError as e:
    print("Erro:", e)

# Tentando inserir elemento -> também dá erro
try:
    animais[0] = "cavalo"
except TypeError as e:
    print("Erro:", e)

# Criando as listas
mamiferos = ["cachorro", "gato", "elefante", "leão"]
insetos = ["formiga", "abelha", "besouro"]

# Criando uma terceira lista a partir delas
animais = mamiferos + insetos
print("Lista de animais:", animais)

# Usando append (funciona)
animais.append("tigre")
print("Após append:", animais)

# Inserindo em uma posição específica
animais.insert(0, "cavalo")
print("Após insert:", animais)
```
